```python
import logging 
from pathlib import Path
import pandas as pd 
import numpy as np 
import joblib 

from sklearn.model_selection import train_test_split 
from sklearn.compose import ColumnTransformer 
from sklearn.preprocessing import StandardScaler , OneHotEncoder , OrdinalEncoder
from sklearn.pipeline import Pipeline 
from sklearn.impute import SimpleImputer

logging.basicConfig(
    level=logging.INFO , 
    format="%(asctime)s [%(levelname)s]: %(message)s"
)

logger = logging.getLogger("preprocessing.ipynb")

pd.set_option("display.max_columns", None)
```

```python
DATA_PATH = Path("../data/processed/featured_supply_chain.csv")

df = pd.read_csv(DATA_PATH)

df.head(5)
```

```python
TARGET_COLUMN = "Late_delivery_risk" 

X = df.drop(columns=[TARGET_COLUMN])

Y = df[TARGET_COLUMN]

    
```

```python
X_train , X_test , Y_train , Y_test = train_test_split(
    X , Y ,
    test_size=0.20 ,
    random_state=42 , 
    stratify= Y
)

X_test.shape
Y_train.shape



```

```python
numerical_features = X_train.select_dtypes(include=["int64", "float64"]).columns.tolist()
categorical_features = X_train.select_dtypes(include=["object" , "category" , "bool" ]).columns.tolist()

low_cardinality_features = [col for col in categorical_features if X_train[col].nunique() <= 15]
high_cardinality_features = [col for col in categorical_features if X_train[col].nunique() > 15]


print(f"Numerical features count: {len(numerical_features)}")
print(f"Low-cardinality categorical features (One-Hot): {len(low_cardinality_features)}")
print(f"High-cardinality categorical features (Ordinal): {len(high_cardinality_features)}")


```

```python
# 1. Pipeline for Numerical features
numerical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
# 2. Pipeline for Low-Cardinality Categorical features
low_cardinality_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False))
])
# 3. Pipeline for High-Cardinality Categorical features
high_cardinality_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("ordinal", OrdinalEncoder(handle_unknown="use_encoded_value", unknown_value=-1))
])  


# Combine all transformers into a single ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numerical_pipeline, numerical_features),
        ("low_card", low_cardinality_pipeline, low_cardinality_features),
        ("high_card", high_cardinality_pipeline, high_cardinality_features)
    ],
    remainder="drop"
)
preprocessor
```

```python
X_train_transformed = preprocessor.fit_transform(X_train)
X_test_transformed = preprocessor.fit_transform(X_test) 

print(X_test_transformed.shape)
print(X_train_transformed.shape)
```

```python
OUTPUT_DATA_DIR = Path("../data/processed")
OUTPUT_MODELS_DIR = Path("../models")


OUTPUT_DATA_DIR.mkdir(parents=True , exist_ok=True)
OUTPUT_MODELS_DIR.mkdir(parents=True , exist_ok=True)

np.savez_compressed(
    OUTPUT_DATA_DIR / "preprocessed_data.npz",
    X_train = X_train_transformed , 
    X_test = X_test_transformed , 
    Y_test = Y_test.to_numpy() , 
    Y_train = Y_train.to_numpy()
 )


joblib.dump(preprocessor , OUTPUT_MODELS_DIR / "preprocessor.joblib")

print("all preprocessed data saved ")
```
